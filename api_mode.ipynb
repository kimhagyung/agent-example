```python
import sseclient
import json 

message = {
    "appName" : APP_NAME,
    "userId" : USER_ID,
    "sessionId" : SESSION_ID,
    "newMessage" : {
        "parts":[ {"text" : "What is the weather there?"} ],
        "role" : "user",
    },
    "streaming" : True  #기본값은 false 이며 true 시에는 토큰 단위로 쪼개어서 출력됨 
} 
response = requests.post(
    f"{BASE_URL}/run_sse", 
    json=message, 
    stream=True
) # run_sse로 변경 (비동기 처리 방식으로 변경, 스트리밍 방식인듯 ?

client= sseclient.SSEClient(response)  # sseclient 생성 

for event in client.events():  # 서버가 보낸 이벤트 반복문 
    data = json.loads(event.data)
    content = data.get('content')
    parts = content.get('parts')

    for part in parts:
        function_call = part.get("functionCall",None)
        if function_call:
            print(function_call.get("name"))
        text = part.get("text",None)
        if text:
            print(text) 
    print("="*60)
```
