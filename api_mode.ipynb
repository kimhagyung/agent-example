```python
import requests

BASE_URL = "http://127.0.0.1:8000"  
APP_NAME = "travel_advisor_agent"
USER_ID = "u_123"
SESSION_ID = "8609013a-cc41-4da9-8e87-5bb988896557"
```

```python
# 서버 재실행하지 않도록 조심하기 (세션 날아가서) 
respnse = requests.post(f"{BASE_URL}/apps/{APP_NAME}/users/{USER_ID}/sessions")

print(respnse.json()) 
```

```python
message = {
    "appName" : APP_NAME,
    "userId" : USER_ID,
    "sessionId" : SESSION_ID,
    "newMessage" : {
        "parts":[
            {"text" : "I want to go to Tokyo."} # Yes, I want to know the currency exchange rate.
        ],
        "role" : "user"
    }
} 
response = requests.post(f"{BASE_URL}/run", json=message)
```

```python
data = response.json()

for event in data:
    content = event.get('content')
    parts = content.get('parts')

    for part in parts:
        function_call = part.get("functionCall",None)
        if function_call:
            print(function_call.get("name"))
        text = part.get("text",None)
        if text:
            print(text) 
    print("="*60)
```
